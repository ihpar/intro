```python
""" 
100cm'den fazla kar yağarsa "Afet"
50 - 100cm arası "Dikkat"
0 - 50cm arası "Normal"
"""

kar = 100

if kar >= 100:
    print("Afet")
if (kar >= 50) and (kar < 100):
    print("Dikkat")
if (kar >= 0) and (kar < 50):
    print("Normal")
```
