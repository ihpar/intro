```python
""" 
if şart:
  t1
  t2 
else:
  f1
  f2
  f3

devam
"""

yas = int(input("Kaç yaşındasın?"))

if yas >= 18: # conditional statement
  print("Ehliyet alabilirsin.")
  print("Oy kullanabilirsin.")
else:
  print("Biraz daha büyümen lazım.")

print("Görüşürüz..")
```

```python
""" 
Bir öğrencinin dönem sonu notu:
vizenin %30'u + finalin %70'i olarak hesaplanır.
Öğrencinin dersi geçmesi için 
dönem sonu notunun en az 55 olması gereklidir.

Öğrencinin vize ve final notunu öğrenciye soran ve dönem sonunda
dersi geçip geçemeyeceğini öğrenciye bildiren bir program yazınız.
"""

vize = int(input("Vize?"))
final = int(input("Final?"))

donem_sonu = (vize * 0.3) + (final * 0.7)
print("Dönem sonu notun:", donem_sonu)
if donem_sonu < 55:
  print("Maalesef hocam...")
else:
  print("Helalyum.")
```
