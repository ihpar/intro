```python
""" 
Öğrencinin notu 55'ten küçükse kötü
[55 - 80) orta
[80 - 100] iyi
Not 0'dan küçük veya 100'den büyükse hata mesajı
"""
notum = 100

if notum >= 0 and notum <= 100:

    if notum < 55:
        print("Kötü")

    if notum >= 55 and notum < 80:
        print("Orta")

    if notum >= 80:
        print("İyi")

else:
    print("Hatalı not girdiniz.")
```
