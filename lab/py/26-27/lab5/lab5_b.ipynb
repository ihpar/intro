```python
"""  
chatbot: q -> çıkış
diğer: kullanıcı mesajı tekrarlanır.
"""
print("Merhaba ben chatbot.")

mesaj = input()

if mesaj == "q":
    print("Görüşürüz...")
else:
    print("Bana", mesaj, "mı dedin?")
    mesaj = input()
    if mesaj == "q":
        print("Görüşürüz...")
    else:
        print("Bana", mesaj, "mı dedin?")
        mesaj = input()
        if mesaj == "q":
            print("Görüşürüz...")
        else:
            print("Bana", mesaj, "mı dedin?")

print("Chatbot sonlandı...")
```

```python
""" 
döngüler: loops
"""

""" 
while şart:
    komutlar
"""

mesaj = input()

while mesaj != "q":
    print("Bana", mesaj, "mı dedin?")
    mesaj = input()


print("Bye...")
```

```python
""" 
1'den 10'a kadarki sayıların toplamı

(...((((1) + 2) + 3) + 4) + ... + 10)
"""

""" 
toplam sayi
0       1
1       2
3       3
6       4
10      5
...     ...
"""

toplam = 0
sayi = 1
while sayi <= 10:
    toplam += sayi # toplam = toplam + sayi
    sayi += 1

print(toplam)
```

```python
"""  
kullanıcıdan aldığı pozitif tam sayının faktöryelini hesaplar.
Ör: kullanıcı 5 girerse 120 hesaplar -> ((((1 * 2) * 3) * 4) * 5)
Ör: kullanıcı 6 girerse 720 hesaplar -> 1 * 2 * 3 * 4 * 5 * 6
"""

sayi = 1
carpim = 1
ust_limit = int(input("Kaç faktöryeli hesaplamak istersin?"))

while sayi <= ust_limit:
    carpim = carpim * sayi 
    # print(sayi, carpim)
    sayi += 1

print(ust_limit, "faktöryel =", carpim)
```
